```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_ollama import ChatOllama

model = ChatOllama(
    model= "qwen-local",
    base_url="http://localhost:11434",
    temperature= 0.3
)

prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system", 
            "تو یک دستیار آموزشی هستی"
        ),
            (
        "human",
        "درباره موضوع {topic} در سطح {level} توضیح بده"
    )
    ]
)

messages = prompt.invoke({
    "topic" : "هوش مصنوعی",
    "level" : "کودکان"
})

response = model.invoke(messages)
print(response.content)
```

```python
### But with LangChain...
chain = prompt | model
response = chain.invoke({
    "topic" : "هوش مصنوعی",
    "level" : "کودکان"
})
print(response.content)
```

```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END

class State(TypedDict):
    name : str  # OR: name:name
    message : str

def create_greet(state: State):
    name = state['name']
    return {
        "message" : f"Hello {name}"
    }

builder = StateGraph(State)
builder.add_node("greeting", create_greet)
builder.add_edge(START, "greeting")
builder.add_edge("greeting", END)
graph = builder.compile()
result = graph.invoke(
    {
        "name" : "Reihaneh"
    }
)

print(result["message"])
```

```python
from langgraph.graph import StateGraph, START, END, MessagesState
from langchain_ollama import ChatOllama

model = ChatOllama(
    model= "qwen-local",
    base_url="http://localhost:11434",
    temperature= 0.3
)

def chatbot(state : MessagesState):
    response = model.invoke(state["messages"])
    return{
        "messages" : [response]
    }

builder = StateGraph(MessagesState)
builder.add_node("chatbot", chatbot)
builder.add_edge(START, "chatbot")
builder.add_edge("chatbot", END)
graph = builder.compile()
result = graph.invoke({
    "messages":[
        (
            "system",
            "تحلیل فروش محصول رو بده"
        ),
        (
            "human",
            "از دسته کتاب‌هایی که تبلیغ فراوانی واسش صورت گرفته و باعث میشه خواننده شتابزده اونو بخره ولی سطحش پایین تر از توقع او خواهد بودالبته کتاب بدی نیست. نویسنده از تجربه زیستش نوشته و لذتی که از سکوت و پیاده روی میبره. و همین دو تجربه زیستی مشابه، حس خوشایندی داشت برای من. ولی اینکه سعی کرده به شکلی ناشیانه جنبه فلسفی بهش بده، باعث شده مفهوم گاهی گنگ بشه.از هایدگر نقل قول کرده ولی ساده ترین معنا رو از اون جمله به ذهن مخاطب میاره. میشه گفت در تلاش برای وزن دادن به کتابش از فلاسفه بزرگ سوءاستفاده کرده!"
        )

    ]
})

print(result["messages"][-1].content)
```
