```python
from hazm import Normalizer, word_tokenize, POSTagger
tagger = POSTagger(repo_id="roshan-research/hazm-postagger", model_filename="pos_tagger.model")

# متن آزمایشی
text = "من امروز یک کتاب خوب خواندم."

# نرمال‌سازی
normalizer = Normalizer()
text = normalizer.normalize(text)

# جدا کردن کلمات
tokens = word_tokenize(text)

print("Tokens:")
print(tokens)

# POS Tagger
# tagger = POSTagger()

# تست برچسب‌گذاری
result = tagger.tag(tokens)

print("\nPOS Tags:")
for word, tag in result:
    print(f"{word:15} -> {tag}")
```
