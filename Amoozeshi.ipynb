```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import hazm
from hazm import Normalizer, word_tokenize, Lemmatizer
import parsivar
```

```python
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)
```

```python
text = "The cat sat on the carpet  because it was tired!"
token = word_tokenize(text)
print(token)
```

```python
text1 = "گربه روی حصیر نشسته   بود چون خسته بود!"
normal = Normalizer()
norm = normal.normalize(text1)
print(norm)
```

```python
tokens = word_tokenize(norm)
print(tokens)
```

```python
lemmatiz = Lemmatizer()
lem = lemmatiz.lemmatize("خوابیده بود")
print(lem)
```
