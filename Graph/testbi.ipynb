```python

import sys, os
from dotenv import load_dotenv
import plotly.graph_objects as go

# ۱. تزریق دقیق متغیرهای محیطی که فایل db.py دوست شما نیاز دارد
os.environ["DB_NAME"] = "postgres"
os.environ["DB_READONLY_USER"] = "postgres"
os.environ["DB_READONLY_PASSWORD"] = "HiddenPatern"
os.environ["DB_HOST"] = "localhost"
os.environ["DB_PORT"] = "5432"

load_dotenv()
sys.path.append(os.path.abspath('.'))
from graph.chart_agent import run_chart_tool   # نه bi_agent / run_bi_tool

def show_my_chart():
    test_sql = """
    SELECT cluster_name, COUNT(user_id) as user_count
    FROM kpi.ml_user_clusters
    GROUP BY cluster_name;
    """
    result = run_chart_tool(
        sql=test_sql, chart_type="bar",
        title="توزیع کاربران در خوشه‌های رفتاری",
        x_field="cluster_name", y_field="user_count",
    )
    if "error" in result:
        print(f"❌ خطا: {result['error']}")
        return
    fig = go.Figure(result["plotly_figure"])
    fig.show()

show_my_chart()
```
